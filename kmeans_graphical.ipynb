```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('../datasets/student_clustering.csv')
```

```python
df.head()
```

```python
df.shape
```

```python
import matplotlib.pyplot as plt
plt.scatter(df['cgpa'],df['iq'])
```

```python
from sklearn.cluster import KMeans
```

```python
from PIL import Image
Image.open("wcss.png")
```

```python
wcss = []

for i in range(1,11):
    km = KMeans(n_clusters=i)
    km.fit_predict(df)
    wcss.append(km.inertia_)
    #Intertia is distance between centroid and the points basically wcss
```

```python
wcss
```

```python
plt.plot(range(1,11),wcss)
```

```python
X = df.iloc[:,:].values
km = KMeans(n_clusters=4)
y_means = km.fit_predict(X)
# y_means now stores the cluster numbers
```

```python
y_means
```

```python
X[y_means == 3,1]
# It shows the df_vals which are of cluster number 3, is show the row 1 out rows 0 and 1
```

```python
plt.scatter(X[y_means == 0,0],X[y_means == 0,1])
plt.scatter(X[y_means == 1,0],X[y_means == 1,1])
plt.scatter(X[y_means == 2,0],X[y_means == 2,1])
plt.scatter(X[y_means == 3,0],X[y_means == 3,1])
```

```python
# Now if it's a 3d data

from sklearn.datasets import make_blobs
centroids = [(-5,-5,5),(5,5,-5),(3.5,-2.5,4),(-2.5,2.5,-4)]
cluster_std = [1,1,1,1]

X,y = make_blobs(n_samples=100,cluster_std=cluster_std,n_features=3,centers=centroids,random_state=42)
```

```python
import plotly.express as px
fig = px.scatter_3d(x = X[:,0], y = X[:,1],z = X[:,2])
fig.show()
```
