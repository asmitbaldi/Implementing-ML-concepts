```python
np.random.seed(12)
X = np.random.rand(1000,1) * 4*np.pi - 2*np.pi
y = np.sin(X[:,0]) + 0.1 * np.random.randn(1000)
```

```python
plt.scatter(X,y,s=5)
```

```python
np.random.seed(42)
n = 100000
X = np.random.rand(n,1)
y = 3*X[:,0]**3 - X[:,0] + np.random.randn(n)
```

```python
plt.scatter(X,y,s=5)
```

```python
np.random.seed(42)
n = 10000
X = np.random.rand(n,2) * 10 - 5
y = (X[:,0]+X[:,1]>0).astype(int)
flip_mask = np.random.rand(n) < 0.1
y[flip_mask] = 1 - y[flip_mask]
```

```python
plt.scatter(X[:,0],X[:,1],s=5,c=y,cmap='coolwarm')
```
