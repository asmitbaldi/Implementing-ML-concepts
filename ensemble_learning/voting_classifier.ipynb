```python
import numpy as np
```

```python
from sklearn.datasets import load_iris
```

```python
X,y = load_iris(return_X_y=True,as_frame=True)
```

```python
X = X[['sepal length (cm)','sepal width (cm)']]
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2,random_state=69)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_val_score
```

```python
lr = LogisticRegression(max_iter=1000)
knn = KNeighborsClassifier()
rf = RandomForestClassifier()
```

```python
estimators = [('lr',lr), ('knn',knn), ('rf',rf)]
```

```python
print("Logistic Regression:",np.round(np.mean(cross_val_score(lr,X,y,cv=10,scoring='accuracy')),2))
print("KNN:",np.round(np.mean(cross_val_score(knn,X,y,cv=10,scoring='accuracy')),2))
print("Random Forest:",np.round(np.mean(cross_val_score(rf,X,y,cv=10,scoring='accuracy')),2))
```

```python
from sklearn.ensemble import VotingClassifier
```

```python
#Let's first try with hard voting
```

```python
vc = VotingClassifier(estimators,voting='hard')
np.round(np.mean(cross_val_score(vc,X,y,cv=10,scoring='accuracy')),2)
```

```python
#Now soft voting
```

```python
vc = VotingClassifier(estimators,voting='soft')
np.round(np.mean(cross_val_score(vc,X,y,cv=10,scoring='accuracy')),2)
```

```python
#Now also adding weights to give more weightage to the good model
```

```python
vc = VotingClassifier(estimators,voting='soft',weights=[5,2,2])
np.round(np.mean(cross_val_score(vc,X,y,cv=10,scoring='accuracy')),2)
```

```python
#now let's try using SVM
```

```python
from sklearn.svm import SVC
```

```python
svm1 = SVC(probability=True,kernel='poly',degree=1)
svm2 = SVC(probability=True,kernel='poly',degree=2)
svm3 = SVC(probability=True,kernel='poly',degree=3)
svm4 = SVC(probability=True,kernel='poly',degree=4)
svm5 = SVC(probability=True,kernel='poly',degree=5)
```

```python
estimators = [('svm1',svm1),('svm2',svm2),('svm3',svm3),('svm4',svm4),('svm5',svm5),]
```

```python
vc = VotingClassifier(estimators,voting='soft')
np.round(np.mean(cross_val_score(vc,X,y,scoring='accuracy',cv=10)),2)
```
