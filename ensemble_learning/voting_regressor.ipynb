```python
from sklearn.datasets import fetch_california_housing
X,y = fetch_california_housing(return_X_y=True)
```

```python
import numpy as np
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.svm import SVR
from sklearn.model_selection import cross_val_score
```

```python
lr = LinearRegression()
dtr = DecisionTreeRegressor()
svr = SVR()
```

```python
estimators = [('lr',lr),('dtr',dtr),('svr',svr)]
```

```python
print("Linear Regression:",np.round(np.mean(cross_val_score(lr,X,y,cv=10,scoring='r2')),2))
print("DecisionTreeRegressor:",np.round(np.mean(cross_val_score(dtr,X,y,cv=10,scoring='r2')),2))
print("Support Vector Regression:",np.round(np.mean(cross_val_score(svr,X,y,cv=10,scoring='r2')),2))
```

```python
from sklearn.ensemble import VotingRegressor
vr = VotingRegressor(estimators)
```

```python
np.round(np.mean(cross_val_score(vr,X,y,cv=10,scoring='r2')),2)
```

```python
dtr1 = DecisionTreeRegressor(max_depth=2)
dtr2 = DecisionTreeRegressor(max_depth=4)
dtr3 = DecisionTreeRegressor(max_depth=6)
dtr4 = DecisionTreeRegressor(max_depth=8)
dtr5 = DecisionTreeRegressor(max_depth=10)
```

```python
estimators = [('dtr1',dtr1),('dtr2',dtr2),('dtr3',dtr3),('dtr4',dtr4),('dtr5',dtr5)]
```

```python
print("DTR1:",np.round(np.mean(cross_val_score(dtr1,X,y,cv=10,scoring='r2')),2))
print("DTR2:",np.round(np.mean(cross_val_score(dtr2,X,y,cv=10,scoring='r2')),2))
print("DTR3:",np.round(np.mean(cross_val_score(dtr3,X,y,cv=10,scoring='r2')),2))
print("DTR4:",np.round(np.mean(cross_val_score(dtr4,X,y,cv=10,scoring='r2')),2))
print("DTR5:",np.round(np.mean(cross_val_score(dtr5,X,y,cv=10,scoring='r2')),2))
```

```python
vr = VotingRegressor(estimators)
np.round(np.mean(cross_val_score(vr,X,y,cv=10,scoring='r2')),2)
```
